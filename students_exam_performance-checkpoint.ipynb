```python
import pandas as pd
```

```python
df = pd.read_csv(r'C:\Users\ASUS\Downloads\pbi projects\students_report\student_exam_performance.csv')
df.info()
df.isnull().sum()
df.describe()
```

```python
df['parent_education'] = df['parent_education'].fillna('Unknown')
df['attendance_percentage'] = df['attendance_percentage'].fillna(df['attendance_percentage'].median())
df['time_management_score'] = df['time_management_score'].fillna(df['time_management_score'].median())
```

```python
df['private_tuition'] = df['private_tuition'].astype(bool)
df['internet_access'] = df['internet_access'].astype(bool)
```

```python
df.drop_duplicates(subset='student_id', inplace=True)
```

```python
df.to_csv('students_cleaned.csv', index=False)
```
